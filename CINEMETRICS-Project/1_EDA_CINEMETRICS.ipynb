```python
df = pd.read_csv('tmdb_1550_movies.csv', encoding='mac_roman')
df.head()
```

```python
df.columns
```

```python
df.drop(columns=['status', 'backdrop_path', 'homepage', 'overview', 'poster_path', 'tagline', 'production_countries',	'spoken_languages'], inplace=True)
```

```python
df = df.rename(columns={
    "runtime": "runtime_minutes",
    "revenue": "revenue_usd",
    "budget": "budget_usd"
})
```

```python
df.head()
```

```python
x = df.isnull().sum()
x[x>0]
```

```python
# DD/MM/YYYY
df["release_year"] = pd.to_datetime(df["release_date"], format="%d/%m/%Y").dt.year
df["release_month"] = pd.to_datetime(df["release_date"], format="%d/%m/%Y").dt.month
```

```python
df.drop(columns=['release_date'], inplace=True)
```

```python
# Drop rows where budget is 0 or NaN
df = df[df["budget_usd"].notna() & (df["budget_usd"] != 0)]
```

```python
# ROI
df.loc[:, "roi_percentage"] = ((df["revenue_usd"] - df["budget_usd"]) / df["budget_usd"]) * 100
```

```python
df.head()
```

```python
df["roi_percentage"] = df["roi_percentage"].round(0)
```

image.png

```python
def roi_to_category(roi):
    if roi == "Unknown":
        return "Unknown"

    if roi < 0:
        return "Flop / Disaster"
    elif roi < 50:
        return "Average / Below Average"
    elif roi < 100:
        return "Hit / Semi-Hit"
    elif roi < 200:
        return "Super Hit"
    elif roi < 400:
        return "Blockbuster"
    else:
        return "All-Time Blockbuster"

df["success_category"] = df["roi_percentage"].apply(roi_to_category)
```

```python
df["roi_percentage"] = df["roi_percentage"].round().astype("Int64")
```

```python
df.head()
```

```python
df.to_csv("movies.csv", index=False) #save csv file
```

The dataset contains extreme values that reflect the heavy-tailed nature of the movie industry rather than data errors. Therefore, outliers were retained and handled using robust techniques such as log transformation and percentile capping (for ROI) instead of removal.

```python
skewed_cols = ['revenue_usd', 'budget_usd', 'vote_count', 'popularity']

df_eda = df.copy()
for col in skewed_cols:
    df_eda[col] = np.log1p(df_eda[col])
```

```python
print(df[skewed_cols].skew())
print("\n")
print(df_eda[skewed_cols].skew())
```

```python
# Handle ROI outliers
low = df['roi_percentage'].quantile(0.01)
high = df['roi_percentage'].quantile(0.99)

df['roi_percentage'] = df['roi_percentage'].clip(low, high)
```

### Count of outliers in numerical columns

```python
def count_outliers_iqr(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()

cols = ['revenue_usd', 'budget_usd', 'vote_count', 'popularity', 'roi_percentage']

for col in cols:
    print(col, ":", count_outliers_iqr(df[col]))
```

Despite applying log transformation and percentile capping, a small number of observations are still flagged by the IQR method. These represent legitimate extreme cases inherent to the movie industry and were intentionally retained to preserve real-world variability.

## Cast Presence vs Movie Success

```python
# Copy dataframe
df_cast = df.copy()

# Split and explode cast names
df_cast['top_casts'] = df_cast['top_casts'].str.split(',')
df_cast = df_cast.explode('top_casts')

# Clean actor names
df_cast['top_casts'] = df_cast['top_casts'].str.strip()

# Keep top 8 most frequent actors to avoid clutter
top_actors = (
    df_cast['top_casts']
    .value_counts()
    .head(8)
    .index
)

df_top = df_cast[df_cast['top_casts'].isin(top_actors)]

# Create count table (actor × success category)
actor_success = pd.crosstab(
    df_top['top_casts'],
    df_top['success_category']
)

# Plot stacked bar chart
actor_success.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7)
)

plt.title('Cast Presence vs Movie Success')
plt.xlabel('Actor')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Success Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
